```python

df = pd.read_csv("/home/wagner-barbosa/Documents/promptcast_forecast/data/processed/guaratuba_processed_1day.csv")

feature_blocks = {}

def add_cyclical_block(df, col_name, max_val, blocks_dict):
    if col_name in df.columns:
        df[f'{col_name}_sin'] = np.sin(2 * np.pi * df[col_name] / max_val)
        df[f'{col_name}_cos'] = np.cos(2 * np.pi * df[col_name] / max_val)
        blocks_dict[col_name] = [f'{col_name}_sin', f'{col_name}_cos']
    return df

df["timestamp"] = (pd.to_datetime(df["timestamp"]))

df = df.drop(columns={"year", "month"})

df = add_cyclical_block(df, "dayofweek", 7, feature_blocks)
df = add_cyclical_block(df, "season", 4, feature_blocks)

df = df.drop(columns={"dayofweek", "season"})

df.to_csv("guaratuba_processed_1day.csv", index=False)
```
